# Notebook 02: Data Preprocessing

Perform justified preprocessing and feature extraction from the date.

```python
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv("../data/raw/instagram_data.csv")
```

## Handle Missing Values and Duplicates

```python
df = df.dropna()
df = df.drop_duplicates()
```

## Parse Dates and Extract Temporal Features
Since exact time is missing, we extract day of week, month, and weekend indicator.

```python
df["Post Date"] = pd.to_datetime(df["Post Date"])
df["day_name"] = df["Post Date"].dt.day_name()
df["day_of_week"] = df["Post Date"].dt.dayofweek # 0=Mon, 6=Sun
df["month"] = df["Post Date"].dt.month
df["is_weekend"] = df["day_of_week"].apply(lambda x: 1 if x >= 5 else 0)
```

## Handle Hashtags

```python
df["Hashtags"] = df["Hashtags"].fillna("")
df["hashtag_count"] = df["Hashtags"].apply(lambda x: len(str(x).split(",")) if str(x).strip() != "" else 0)
```

## Save Processed Dataset

```python
os.makedirs("../data/processed", exist_ok=True)
df.to_csv("../data/processed/processed_data.csv", index=False)
print("Shape of processed data:", df.shape)
```
